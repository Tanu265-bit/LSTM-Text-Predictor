```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([faqs])
```

```python
len(tokenizer.word_index)
# give index no. of words tokenizer.word_index
```

```python
input_sequences = []
for sentence in faqs.split('\n'): # all sentences n faq
  tokenized_sentence = tokenizer.texts_to_sequences([sentence])[0]
  #tokenizer.texts_to_sequences([sentence]) - convert words to no.
  # tokenizer.texts_to_sequences([sentence])[0] -0 se sb sentences 1 br ayege bs

  for i in range(1,len(tokenized_sentence)):# loop 1 se toeknizer ki length tk jayege mean 1-3
  #:i+1]) slicing - all to i+1 tk loop chlega mean 1-2
    input_sequences.append(tokenized_sentence[:i+1])
    print("Number of sequences:", len(input_sequences))
```

```python
input_sequences
```

```python
max_len = max([len(x) for x in input_sequences])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input_sequences = pad_sequences(input_sequences, maxlen = max_len, padding='pre')
print("Max length:", max_len)
print("Padded shape:", padded_input_sequences.shape)
# 0 phle lgana chate h padding='pre'
# 0 padding mean jo needed ni hoge /missing hoge unhe 0 likh dege
```

```python
padded_input_sequences
```

```python
X = padded_input_sequences[:,:-1]# all to left last column
```

```python
y = padded_input_sequences[:,-1]#all row nd last col
```

```python
X.shape
```

```python
y.shape
```

```python
from tensorflow.keras.utils import to_categorical
# utility strt from 0 nd totl length was 282 so we wrote 283
y = to_categorical(y,num_classes=283)
```

```python
y.shape
```

```python
print(X.shape)
```

```python

# o/p dence vector me 100 no. hoge,nd hr row me 56 input words hoge
#gibve dense vector for every vector
# 150 nodes
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, LSTM, Dense

vocab_size = len(tokenizer.word_index) + 1

model = Sequential()

model.add(Input(shape=(56,)))

model.add(Embedding(
    input_dim=vocab_size,
    output_dim=100,
    mask_zero=True
))

model.add(LSTM(150, return_sequences=True))

model.add(LSTM(150))

model.add(Dense(
    vocab_size,
    activation='softmax'
))
```

```python
model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
history = model.fit(
    X,
    y,
    epochs=100
)
```

```python
print("Final accuracy:", history.history['accuracy'][-1])
print("Final loss:", history.history['loss'][-1])
```
